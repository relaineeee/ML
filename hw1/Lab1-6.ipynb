4. Как найти самое часто встречаемое значение между двумя массивами? 

````python
a = np.array([1, 2, 3, 3, 4])
b = np.array([2, 3, 3, 3, 4])
c = np.concatenate((a, b))
values, counts = np.unique(c, return_counts=True)
result = values[np.argmax(counts)]
print(result)
````

Motivation_Level - уровень мотивации

````python
import pandas as pd
````

````python
df = pd.read_csv('student_perfomance.csv')
df.head()
````

1. Посчитайте среднее кол-во часов уделяемое на сон студентами

````python
print(f'{df['Sleep_Hours'].mean():.2f} ч.')
````

2. Сколько процентов студентов посещают внеклассовые активности?

````python
print(f'{(df['Extracurricular_Activities'] == 'Yes').mean() * 100:.2f}%')
````

3. Правда ли, что в среднем девушки сдают экзамены лучше, чем мужчины

````python
a = df[df['Gender'] == 'Female']['Exam_Score'].mean()
b = df[df['Gender'] == 'Male']['Exam_Score'].mean()
print(a > b)
````

4. Выведите 25 и 75 перцентили времени сна студентов

````python
print(df['Sleep_Hours'].quantile([0.25, 0.75]))
````

5. Правда ли, что те, кто спит больше(75 перцентиль и выше) сдает экзамен лучше, чем студенты с недосыпом(25 перцентиль и меньше)

````python
q25 = df['Sleep_Hours'].quantile(0.25)
q75 = df['Sleep_Hours'].quantile(0.75)
a = df[df['Sleep_Hours'] >= q75]['Exam_Score'].mean()
b = df[df['Sleep_Hours'] <= q25]['Exam_Score'].mean()
print(a > b)
````

6. Влияет ли физическая активность на результаты экзамена?

````python
print(df.groupby('Physical_Activity')['Exam_Score'].mean())
print('Заметной связи нет')
````

7. Правда ли, что слуденты, чьи родители имеют высшее образование(College || Postgraduate), сдают экзамены лучше, чем остальные?

````python
a = df[df['Parental_Education_Level'].isin(['College', 'Postgraduate'])]['Exam_Score'].mean()
b = df[df['Parental_Education_Level'] == 'High School']['Exam_Score'].mean()
print(a > b)
````

8. Как влияет участие родителей на результаты экзаменов

````python
print(df.groupby('Parental_Involvement')['Exam_Score'].mean())
print('Чем выше участие родителей, тем выше средний балл')
````

9. Правда ли, что в семьях с доходом больше среднего дети сдают экзамены лучше?

````python
a = df[df['Family_Income'] == 'High']['Exam_Score'].mean()
b = df[df['Family_Income'].isin(['Low', 'Medium'])]['Exam_Score'].mean()
print(a > b)
````

10. Посчитайте долю студентов сдавших экзамен на более чем 70 баллов в семьях с низким доходом и отсутствием доступа к интернету

````python
a = df[(df['Family_Income'] == 'Low') & (df['Internet_Access'] == 'No')]
b = (a['Exam_Score'] > 70).mean()
print(f'{b} ({b * 100:.2f}%)')
````

### 3. Визуализация

Используя библиотеки matplotlib/sns/plotly/pandas и набор данных представленный ранее визуализируйте

1. Правда ли, что количество часов затрачиваемых на сон распределено примерно нормально?

````python
import matplotlib.pyplot as plt
import seaborn as sns
sns.histplot(data=df, x='Sleep_Hours', bins=7, kde=True)
plt.title('Распределение времени сна')
plt.xlabel('Часы сна')
plt.ylabel('Количество студентов')
plt.grid(True, alpha=0.3)
plt.show()
print('Правда')
````

2. Постройте гистограмму распределения количества часов затрачиваемых на обучение среди мальчиков и девочек

````python
sns.histplot(data=df, x='Hours_Studied', hue='Gender', bins=20, kde=True, alpha=0.4)
plt.title('Распределение времени обучения')
plt.xlabel('Часы обучения')
plt.ylabel('Количество студентов')
plt.grid(True, alpha=0.3)
plt.show()
````

3. Как зависит результат конечного экзамена от результата первого экзамена

````python
sns.regplot(data=df, x='Previous_Scores', y='Exam_Score', scatter_kws={'alpha': 0.2}, line_kws={'color': 'red'})
plt.title('Зависимость итогового балла от первого экзамена')
plt.xlabel('Балл за первый экзамен')
plt.ylabel('Итоговый балл')
plt.grid(True, alpha=0.3)
plt.show()
corr = df['Previous_Scores'].corr(df['Exam_Score'])
print(f'Корреляция: {corr:.3f}')
print('Слабая положительная зависимость')
````

4. Постройте box plot отображающий распределение количества сна детей сдавших экзамены на оценку ниже и выше 85 баллов

````python
a = df[df['Exam_Score'] < 85]['Sleep_Hours']
b = df[df['Exam_Score'] > 85]['Sleep_Hours']

plt.boxplot([a, b], tick_labels=['Ниже 85', 'Выше 85'])

plt.title('Распределение времени сна')
plt.xlabel('Результат экзамена')
plt.ylabel('Часы сна')
plt.grid(axis='y', alpha=0.3)
plt.show()
````

5. Правда ли, что количество посещаемых занятий с репетитором в месяц напрямую влияет на балл в конечном экзамене?

````python
a = df.groupby('Tutoring_Sessions')['Exam_Score'].mean()
plt.plot(a.index, a.values, marker='o')
plt.title('Зависимость результата от занятий с репетитором')
plt.xlabel('Количество занятий с репетитором')
plt.ylabel('Средний балл экзамена')
plt.grid(True, alpha=0.3)
plt.show()
corr = df['Tutoring_Sessions'].corr(df['Exam_Score'])
print(f'Корреляция: {corr:.3f}')
print('Слабая положительная зависимость')
````

6. Как физическая активность влияет на успеваемость?

````python
a = df.groupby('Physical_Activity')['Exam_Score'].mean()
plt.bar(a.index, a.values)
plt.title('Физическая активность и успеваемость')
plt.xlabel('Физическая активность')
plt.ylabel('Средний балл экзамена')
plt.grid(axis='y', alpha=0.3)
plt.show()
corr = df['Physical_Activity'].corr(df['Exam_Score'])
print(f'Корреляция: {corr:.3f}')
print('Зависимость практически отсутствует')
````
